```python
#Let's Just make a clear dataframe to find how many movie id are there

movid_id=None
movie_np=[]  #Empty list to store all the movie id

#Iterate over the cust_Id column
for i in netflix_ds['Cust_ID']:
  if ':' in i:
    #Update the current movie id in column
    movie_id=int(i.replace(':',''))
  movie_np.append(movie_id)    #Adding the movie id number in the list
```

```python
#Pre Filtering:
#Remove all the users that have raetd less movies
#Remove all the movies that have rated less
```

```python
#Count the ratings of movie_id
movie_summary=netflix_ds.groupby('Movie_ID')['Ratings'].agg(['count'])
```

```python
movie_summary
```

```python
#Now we will create a benchmark(conisder a benchmark value like 60 percentile)
movie_benchmark=round(movie_summary['count'].quantile(0.60),0)
movie_benchmark
```

```python
#If any movie is having ratings count less than 908 exclude that movie
drop_movie_list=movie_summary[movie_summary['count']<movie_benchmark].index
drop_movie_list
```

```python
#no of removed movies
len(drop_movie_list)
```

```python
#Count how many users have how many rating count
cust_summary=netflix_ds.groupby('Cust_ID')['Ratings'].agg(['count'])
cust_summary
```

```python
#Set the becnhmark
cust_benchmark=round(cust_summary['count'].quantile(0.60),0)
cust_benchmark
```

```python
#If any user have rated less than 36 movies we will exclude that user
drop_cust_list=cust_summary[cust_summary['count']<cust_benchmark].index
```

```python
drop_cust_list
```

```python
len(drop_cust_list)
```

```python
netflix_ds=netflix_ds[~netflix_ds['Movie_ID'].isin(drop_movie_list)]
netflix_ds=netflix_ds[~netflix_ds['Cust_ID'].isin(drop_cust_list)]
```
